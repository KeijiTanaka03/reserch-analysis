```python
import os
```

```python
data_folder = "../data/20250630"

file_paths =[
    "20250630_213940_data.txt",
    "20250630_214428_data.txt",
    "20250630_214843_data.txt",
    "20250630_215347_data.txt",
    "20250630_220021_data.txt",
    "20250630_220508_data.txt",
    "20250630_220953_data.txt",
    "20250630_221442_data.txt",
    "20250630_222042_data.txt",
    "20250630_222537_data.txt"
]

row_data = []
```

```python
# row_dataリストにデータを格納する
# row_data[ラベル番号][ファイル番号][各行のCSIデータ]
# 各行のデータ：{'no': 行番号, 'ms': ミリ秒, 'csi': CSIデータ(10進数)}

def parse_csi_line(line):
    parts = line.strip().split(',')
    no = int(parts[0])
    ms = int(parts[1])
    csi_hex = parts[2]
    csi = [int(csi_hex[i:i+2], 16) for i in range(0, len(csi_hex), 2)]
    return {'no': no, 'ms': ms, 'csi': csi}


for i in range(1, 43):  # 1~42
    folder_data = []
    folder = os.path.join(data_folder, str(i))
    for fname in file_paths:
        file_data = []
        fpath = os.path.join(folder, fname)
        with open(fpath) as f:
            for line in f:
                file_data.append(parse_csi_line(line))
        folder_data.append(file_data)


    row_data.append(folder_data)
```

```python
# データの平均サンプリングレートを求める


sampling_rates = []  # サンプリングレートを集めるリスト

for label_idx in range(len(row_data)):
    for file_idx in range(len(row_data[label_idx])):
        # 現在のファイルのデータを取得
        data_rows = row_data[label_idx][file_idx]

        # ms値のリストを作成
        ms_list = [row['ms'] for row in data_rows]

        # ms間の差分（連続する時刻の差）
        diffs = [ms_list[i+1] - ms_list[i] for i in range(len(ms_list) - 1)]

        # サンプリング周期とレートの計算
        if diffs:
            avg_period_ms = sum(diffs) / len(diffs)
            sampling_rate = 1000 / avg_period_ms
            sampling_rates.append(sampling_rate)  # リストに追加
            print(f"ラベル{label_idx}, ファイル{file_idx} → 平均周期: {avg_period_ms:.2f} ms, サンプリングレート: {sampling_rate:.2f} Hz")
        else:
            print(f"ラベル{label_idx}, ファイル{file_idx} → データ不足（サンプリングレート計算不可）")

# 全体の平均サンプリングレートを表示
if sampling_rates:
    overall_avg_rate = sum(sampling_rates) / len(sampling_rates)
    print(f"\n✅ 全データにおける平均サンプリングレート: {overall_avg_rate:.2f} Hz")
else:
    print("\n⚠️ 有効なサンプリングレートが1つも取得できませんでした。")
```
